# Python Assignment 2: Data Structures — List, Dictionary, Set & Conditional Statements

## 1. List (Creation, Modification and Access)

### 1(a–b). Create the lists

```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Bala", "Charan", "Divya", "Eshan"]

print("age_list:", age_list)
print("name_list:", name_list)
```

### 2(a–g). List operations and modifications

```python
# a. Append a name
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2
age_list.insert(2, 30)
print("After inserting 30 at index 2:", age_list)

# c. Remove Yazhini
name_list.remove("Yazhini")
print("After removing Yazhini:", name_list)

# d. Pop the last age
popped_age = age_list.pop()
print("Popped age:", popped_age)
print("After pop:", age_list)

# e. Extend with additional ages
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort in descending order
age_list.sort(reverse=True)
print("Sorted in descending order:", age_list)

# g. Maximum, minimum, and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

### 3(a–d). Accessing list elements

```python
print("First name:", name_list[0])
print("Last name:", name_list[-1])
print("Elements from index 2 to index 4:", name_list[2:5])
print("Names in reverse order:", name_list[::-1])
```

## 2. Dictionary (Creation, Modification and Access)

```python
# a. Map five students to marks (0–100)
student_marks = {
    "Anu": 76,
    "Bala": 91,
    "Charan": 68,
    "Divya": 85,
    "Eshan": 73
}
print("Initial student marks:", student_marks)

# b. Access the mark of a specific student
print("Bala's mark:", student_marks["Bala"])

# c. Add Janani with a mark of 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# d. Update one older student's mark to 82
student_marks["Charan"] = 82
print("After updating Charan:", student_marks)

# e. Print keys, values, and key-value pairs
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

## 3. Sets (Operations)

### a–b. Create a set and examine its behavior

```python
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print("my_set:", my_set)
print("Number of unique elements:", len(my_set))
```

**Explanation:** A set stores unique elements, so repeated values such as `'a'` and `'i'` appear only once. Sets are unordered, so the display order is not guaranteed.

A set does not support indexing. The next cell demonstrates the requested assignment and catches the error so the notebook can continue.

```python
try:
    my_set[4] = 's'
except TypeError as error:
    print("Error:", error)
    print("Explanation: Sets are unordered and do not support item assignment or index-based access.")
```

### c–d. Union and intersection

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("set1:", set1)
print("set2:", set2)
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
