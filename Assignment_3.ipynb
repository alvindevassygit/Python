## 1. While Loop & Control Statements — Number Guessing Game

The game selects a random number from 1 to 10 and allows up to 3 valid guesses. An out-of-range guess does not use an attempt, matching the sample output's behavior.

```python
import random

# Generate a random number between 1 and 10
secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
    try:
        guess = int(input("Guess the number (between 1 and 10): "))
    except ValueError:
        print("Please enter a whole number.")
        continue

    # Out-of-range guesses do not count as an attempt
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    attempts -= 1

    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess > secret_number:
        print("Too high. Try again.")
    else:
        print("Too low. Try again.")
else:
    print(f"Better luck next time! The correct number was {secret_number}.")
```

## 2. For Loop — Multiplication Table Generator

```python
try:
    number = float(input("Enter the number for which you want the multiplication table: "))

    for i in range(1, 11):
        result = number * i
        # Display whole-number results without a decimal point
        if result.is_integer():
            result = int(result)
        display_number = int(number) if number.is_integer() else number
        print(f"{display_number} x {i} = {result}")
except ValueError:
    print("Please enter a valid number.")
```

## 3. Function — BMI Calculator

```python
def calculate_bmi(weight, height):
    """Return BMI calculated from weight in kg and height in meters."""
    if height <= 0:
        raise ValueError("Height must be greater than zero.")
    return weight / (height ** 2)


try:
    weight = float(input("Enter your weight in kg: "))
    height = float(input("Enter your height in meters: "))

    if weight <= 0:
        print("Weight must be greater than zero.")
    else:
        bmi = calculate_bmi(weight, height)
        print(f"Your BMI is: {bmi:.2f}")
except ValueError as error:
    print(f"Invalid input: {error}")
```
